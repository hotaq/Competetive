## Competition Overview
 
The scoring formula is `max(1, 25 − ln(cost))` where `cost = params + bytes + MACs`.
 
| Architecture | Typical cost | Score per task |
|---|---|---|
| Static Identity (cost=0) | 0 | **25.0** |
| Static Channel Gather | ~45–54 | **21.0–21.2** |
| Slice/Pad/Tile chains | ~200–500 | **19–21** |
| Spatial Gather (pixel remap) | ~4,563 | **16.6** |
| 1×1 Conv (static 30×30) | ~90,500 | **13.6** |
| 3×3 Conv (trained) | ~814,500 | **11.4** |
| 2-layer CNN | ~1.4M–2.6M | **10.8** |
 
**Critical rule:** Static shapes `[1,10,30,30]` required. Dynamic shapes, Loop, Scan, NonZero, Unique ops are all banned and score 0.
 
**Critical discovery:** Lower local `macs+memory+params` does **not** always produce a lower Kaggle LB score. Per-task LB-bisection is the only reliable verification method.

## Approach
 
### Primary Strategy — Pre-built Dataset (one-click)
 
The fastest path: attach the `konbu17/neurogolf-2026-blended-401-v117` dataset as input. It contains 401 hand-verified ONNX files (one per task including task000). The notebook then:
 
1. Finds all `task*.onnx` files recursively under `/kaggle/input/`
2. Zips them into `submission.zip`
3. Submit
Total runtime: under 30 seconds.
 
### How the Dataset Was Built
 
The 401 ONNX files are the result of **best-of-N picks with LB-bisection**:
 
- Collect candidate models for each task from multiple upstream sources
- For each swap candidate: run LB submission, keep only if actual LB score improves
- Never trust local `onnx_tool` cost estimate alone — always verify on leaderboard
- Include hand-crafted ONNX for tasks where rule analysis reveals simpler structure
**Upstream sources combined:**
- `afr1ste/neurogolf-5060-75-submission-notes` — strongest learned-conv ladders for many tasks
- `vyankteshdwivedi/neurogolf-multi-source-onnx-solver` — extreme op-fusion/minimization
- `jonathanchan/ngc26-constraint-smart-logic-mix-blending` — constraint-smart logic blend
- `artemnazemtsev/neurogolf-logic-driven-ensembling-part-4` — logic-driven compression
- `thisray/neurogolf-4743-93-submission-task-table` — agent-AI driven task exploration
- Grand Master discussion #694628 — reshape(-1, 3) data-structure hack for task007
### Manual ONNX Rescue
 
For tasks where no upstream source produced a good model, hand-built static ONNX graphs encode the transformation rule directly. Example: task221 uses a tile-grid dispatch by `K=count(non-bg)` with zero learned parameters.
 
---
 
## Key Findings
 
**LB-bisection is the only reliable evaluation method.** Local `onnx_tool` cost (macs+memory+params) frequently disagrees with the official Kaggle scorer due to implementation differences in how operations are counted. A model that appears 2× cheaper locally can score worse on LB. Always verify swaps on the real leaderboard.
 
**The community ecosystem is the moat.** The gap between a 4000-score solo effort and 5385+ comes entirely from combining insights across 6+ public notebooks, each contributing a few dozen task-specific improvements.
 
**task000 exists.** The competition nominally has 400 tasks numbered 1–400, but a task000 is present in the dataset and scores on the LB.
 
**Cost=0 is the ceiling.** task179, task241, and others with pure Slice/Pad operations and no initializers score 25.0. Identifying these patterns at scale is what separates 5000 from 6000+.
 
**Manual rescue scales to hard tasks.** Building a static computation graph that directly encodes a transformation rule (as in task076, task133, task221) recovers tasks that trained models cannot generalize on. Each rescue is worth 10–14 pts.

````python
import os, zipfile, glob

# Search for task*.onnx anywhere under /kaggle/input/
all_onnx = glob.glob('/kaggle/input/**/task*.onnx', recursive=True)
print(f'found {len(all_onnx)} task*.onnx files via recursive glob')

# Also try /kaggle/input/*/submission.zip (in case dataset preserved as zip)
zip_candidates = glob.glob('/kaggle/input/**/submission.zip', recursive=True)
print(f'found {len(zip_candidates)} submission.zip files: {zip_candidates}')

DST = '/kaggle/working/submission.zip'

if zip_candidates:
    # Use the pre-built zip directly
    import shutil
    src = max(zip_candidates, key=os.path.getsize)
    shutil.copyfile(src, DST)
    print(f'\nCopied: {src} -> {DST}')
elif all_onnx:
    # Build zip from .onnx files
    with zipfile.ZipFile(DST, 'w', zipfile.ZIP_DEFLATED) as z:
        for f in sorted(all_onnx):
            z.write(f, arcname=os.path.basename(f))
    print(f'\nZipped {len(all_onnx)} files -> {DST}')
else:
    raise RuntimeError('No task*.onnx or submission.zip found in /kaggle/input/. Make sure the konbu17/neurogolf-2026-blended-401-v117 dataset is attached.')

print(f'Output size: {os.path.getsize(DST):,} bytes ({os.path.getsize(DST)/1024/1024:.2f} MB)')
with zipfile.ZipFile(DST) as z:
    names = sorted(z.namelist())
    print(f'Files in zip: {len(names)}')
    print(f'First 3: {names[:3]}')
    print(f'Last 3: {names[-3:]}')
````
