```python
import zipfile
from pathlib import Path

# Пути
INPUT_DIR = Path('/kaggle/input/datasets/artemnazemtsev/6214-submission')
OUTPUT_ZIP = Path('/kaggle/working/submission.zip')

# Находим все .onnx файлы рекурсивно
onnx_files = list(INPUT_DIR.rglob('*.onnx'))

if not onnx_files:
    print("⚠️ Файлы .onnx не найдены! Проверьте путь.")
else:
    with zipfile.ZipFile(OUTPUT_ZIP, 'w', zipfile.ZIP_DEFLATED) as zf:
        for file_path in onnx_files:
            # Если соревнование требует файлы в корне архива:
            zf.write(file_path, file_path.name)
            
            # 💡 Если нужно сохранить структуру папок внутри архива, замените на:
            # zf.write(file_path, file_path.relative_to(INPUT_DIR))
            
    print(f"✅ Архив создан: {OUTPUT_ZIP.name}")
    print(f"📦 В архиве: {len(onnx_files)} файлов")
```
