## 1. Problem framing

Kaggriculture is not a static tabular prediction problem. Every action changes
shared prices, cash flow, travel feasibility, feed availability, and the value
of future actions. A locally attractive action can therefore invalidate the
rest of a route.

The v15 design follows three hypotheses:

1. **Route coherence dominates local cleverness.** A replay-derived route that
survives many seeds is more valuable than a turn-by-turn policy that frequently
changes its movement plan.
2. **Public portfolio exposure is useful, private inventory is not available.**
Visible crops, animals, ready yields, and unfed animals provide a noisy estimate
of near-future market pressure.
3. **Adaptation should have a narrow action surface.** If evidence is weak, use
the base route. If evidence is strong, reorder only non-WHEAT sales; preserve
the WHEAT cash-and-feed loop.

The consensus gate is evaluated at the start of a day and held for 24 steps. This
hysteresis prevents the policy from oscillating because of one noisy observation.
The expert is not allowed to add a speculative wheat squeeze in the submitted
configuration.

## 3. The observable signal

For each visible opponent asset, v15 builds a soft exposure estimate. A ready
yield receives extra weight; recurring animals retain a smaller cadence weight.
WHEAT exposure also includes animal count because feed demand is the important
buy-side pressure. The classifier then computes:

\[\text{confidence} = \min(1, E/6) \times \min(1, C/0.50)\]\n
where `E` is total non-WHEAT exposure and `C` is the largest product's share of
that exposure. v15 computes this physical confidence and a second confidence
after normalizing by equilibrium price, then takes their minimum. A floor-priced
crop can therefore veto a false collision signal; price alone cannot activate
the expert.

````python
# A small executable illustration of the v15 consensus gate.
def gate_confidence(exposure, prices=None, scale=6.0, target_concentration=0.50):
    values = [max(0.0, float(exposure.get(k, 0.0))) for k in
              ('CARROT', 'TOMATO', 'STRAWBERRY', 'MELON', 'EGG', 'MILK', 'WOOL')]
    total = sum(values)
    concentration = max(values, default=0.0) / total if total else 0.0
    physical = min(1.0, total / scale) * min(1.0, concentration / target_concentration)
    reference = {'CARROT': 35, 'TOMATO': 60, 'STRAWBERRY': 120, 'MELON': 250,
                 'EGG': 50, 'MILK': 160, 'WOOL': 200}
    prices = prices or {}
    weighted = [value * max(1.0, float(prices.get(k, reference[k]))) / reference[k]
                for k, value in zip(reference, values)]
    weighted_total = sum(weighted)
    value_concentration = max(weighted, default=0.0) / weighted_total if weighted_total else 0.0
    value = min(1.0, total / scale) * min(1.0, value_concentration / target_concentration)
    return round(min(physical, value), 3), round(concentration, 3), round(value_concentration, 3)

examples = {
    'balanced farm': {'MILK': 1.5, 'WOOL': 1.2, 'STRAWBERRY': 1.0},
    'concentrated milk farm': {'MILK': 7.0, 'WOOL': 0.4},
    'small signal': {'MILK': 2.0, 'WOOL': 0.2},
}
for name, exposure in examples.items():
    confidence, concentration, value_concentration = gate_confidence(exposure)
    mode = 'COLLISION' if confidence >= 0.70 else 'BASE'
    print(f'{name:24s} confidence={confidence:.3f} physical_C={concentration:.3f} value_C={value_concentration:.3f} -> {mode}')
````

## 4. Validation protocol

I separate selection evidence from generalization evidence:

- **Current top-10 replay league:** 92 train games and a chronological 50-game
  holdout. The holdout is kept out of the gate design decision.
- **Ablation:** compare the v15 consensus gate with the v13 physical gate and
  the market gate.
- **Direct regression check:** 10 seeds × both player positions against v12.
- **Top-five replay check:** one train replay for each current LB top-five team,
  evaluated under identical seeds for v13, v14, and v15.
- **Stress league:** 14 synthetic and historical strategy families, 2 seeds ×
  both positions.

The primary metric is win rate, but I also track mean log score ratio, worst
opponent win rate, minimum reward, and lower-tail CVaR. This avoids selecting a
fragile policy that wins often but catastrophically fails against one style.

## 5. What changed from the core?

The fixed core is the modal action trajectory extracted from senkin13 episode
89634061; it appeared identically in 8/8 selected training trajectories. v15
does **not** replace farmer movement, crop placement, livestock routing, or the
WHEAT working-capital loop.

When both gate views agree, it ranks existing sale orders by:

\[\text{collision value}(p) = \text{visible exposure}(p) \times \text{price}(p) \times \min(\text{quantity}, 10)\]\n
Only non-WHEAT sells with at least 2 units of visible exposure and a price
above the $1 floor can move ahead. A floor-priced crop is deliberately left
in the base order, which is the main defense against a misleading top-five
public portfolio snapshot.
This preserves the part of the route most sensitive to feed-price timing while
still using the shared market as a competitive resource.

## 6. Reproducibility and limitations

The exact self-contained submission artifact is the repository's `main.py`. It
has no opponent-name lookup and uses only observations exposed by the
environment. The repository also contains the validation scripts and frozen
reports used to produce the summary above.

Known limitations:

- Replay opponents do not react to counterfactual actions, so replay wins are
  evidence of robustness, not a substitute for the official tournament.
- The current leaderboard changes over time; the top-10 snapshot must be
  refreshed before a new submission.
- Synthetic stress opponents are weaker and more regular than the strongest
  live competitors.
- A silver or gold notebook medal depends on reader engagement and Kaggle's
  notebook evaluation; the design cannot guarantee a medal.

The practical takeaway is simple: in a stateful competitive environment, a
small, interpretable, delayed gate can be safer than a larger adaptive policy.
v15 is an attempt to make that trade-off explicit and testable.
