### Read the programme from the backbone outward

<div class="kg-roadmap"><div class="kg-road"><span class="n">01</span><span class="t">Keep the backbone</span><span class="d">The supplied native route tapes remain byte-for-byte unchanged.</span></div><div class="kg-road teal"><span class="n">02</span><span class="t">Route on public state</span><span class="d">One late decision uses the first shop and visible market or field state.</span></div><div class="kg-road gold"><span class="n">03</span><span class="t">Change one edge</span><span class="d">Only the last executable turn receives a bounded settlement rule.</span></div><div class="kg-road violet"><span class="n">✓</span><span class="t">Build locally</span><span class="d">The Notebook writes the agent and submission archive without an external Dataset.</span></div></div>

## 1. Why keep the route fixed?

Shop unlocks are sparse, high-value milestones. Between them, a stable movement and production programme avoids needless reactions to small market noise. The supplied agent therefore keeps two long routes and changes route only once, after enough public state is visible.

## 2. The strategy in one line

> Follow the guarded route → read the shop state → choose one tail → settle the final inventory

The policy is intentionally conservative: the route programme does the work, while the new rule touches only inventory that would otherwise remain after the final action.

## 3. One deliberately small strength change

<div class="kg-note gold"><div class="title">Last executable turn only</div><div class="body">Fill spare order slots with SELL orders for products not already covered. The engine resolves unit work before market orders, so a final collection or harvest can create inventory after the original order was planned. An oversized SELL is safe here because the market fills only the available quantity.</div></div>

## 4. Public-state route selection

The original decision tree is preserved. It never reads a seed, player identity, hidden opponent information, or a future shop.

## 5. Self-contained agent materialization

The next cell reconstructs a standalone `main.py` with the native policy embedded inside it, verifies its SHA-256 hash, and creates a ready-to-save `.tar.gz` bundle. Selecting `main.py` by itself is supported; no sidecar file, external Kaggle Dataset, or internet access is required.

```python
import base64, hashlib, tarfile
from pathlib import Path

VARIANT = 'terminal_product_sweep'
EXPECTED_MAIN_SHA256 = 'f9ab87f380da2fc095be528df6c150d51a44646fd7dbe5edbff557ecd64f8af2'
MAIN_B64 = ''

Path("main.py").write_bytes(base64.b64decode(MAIN_B64))

main_sha = hashlib.sha256(Path("main.py").read_bytes()).hexdigest()
assert main_sha == EXPECTED_MAIN_SHA256

archive = Path(f"kaggriculture_{VARIANT}.tar.gz")
with tarfile.open(archive, "w:gz") as bundle:
    bundle.add("main.py", arcname="main.py")

print("agent entry point: main.py")
print("native policy: embedded inside main.py")
print("submission bundle:", archive)
print("main.py SHA-256:", main_sha)
print("external Dataset required: NO")
```
