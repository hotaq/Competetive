## Search objective

Candidates are ordered lexicographically by:

1. every simulation completing successfully;
2. worst opponent-family match points;
3. family-balanced mean match points;
4. mean score margin clipped to ±5000.

This prevents a large win over one weak family from hiding a systematic loss to another. Every match is run from both seats. Development opponents/seeds are eliminated before a separate final round using four families and two untouched seeds.

## Discrete decisions tested

- route family and its implied crop/land portfolio;
- budget, storage, dead-stock, and terminal liquidation safeguards;
- pre-overflow selling and exact market-order response horizons;
- native versus yarn-conditioned herd substitution;
- original, early-wheat, hybrid-wheat, and tomato-for-cow openings;
- tomato watering skip schedule, wheat sale floor, and reservation horizon.

The tomato-for-cow opening failed decisively. No tested mutation improved the reference on the held-out set, so V61 retains the robust reference instead of promoting a noisy or neutral change. Its held-out family points were `{'herd_safe': 0.5, 'measured_v60': 1.0, 'moon_v1': 1.0, 'route_v44': 1.0}` over `16` games.

```python
import gzip, hashlib, io, tarfile
members = [('main.py', MAIN.read_bytes()), ('LICENSE.txt', LICENSE.read_bytes()), ('NOTICE.txt', NOTICE.read_bytes())]
ARCHIVE = OUTPUT_ROOT / 'submission.tar.gz'
tar_buffer = io.BytesIO()
with tarfile.open(fileobj=tar_buffer, mode='w') as tar:
    for name, data in members:
        info = tarfile.TarInfo(name)
        info.size = len(data)
        info.mtime = 0
        info.mode = 0o644
        tar.addfile(info, io.BytesIO(data))
with ARCHIVE.open('wb') as stream:
    with gzip.GzipFile(fileobj=stream, mode='wb', mtime=0, filename='') as zipper:
        zipper.write(tar_buffer.getvalue())
with tarfile.open(ARCHIVE) as tar:
    assert tar.getnames() == [name for name, _ in members]
    for name, data in members:
        assert tar.extractfile(name).read() == data
print('Ready:', ARCHIVE)
print('main.py SHA256:', EXPECTED_MAIN_SHA256)
```
