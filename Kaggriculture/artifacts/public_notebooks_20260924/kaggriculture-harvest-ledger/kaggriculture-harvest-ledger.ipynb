# V64 — Harvest Ledger — Demand-Preserving Sales

Reproduce the credited tetsutani controller with its full inherited production and market stack. Its final layer activates from turn 216 and reorders only contiguous sale blocks while preserving quantities and non-sale slots. The underlying composite includes additional inherited timing and seed-budget layers; this is not merely a one-line patch to V57. A clone-based price model proposes changes, so modeled gains do not guarantee gains against real opponents.

This is a credited public-strategy refresh. These two public branches were selected on screening seeds, then checked on disjoint confirmation seeds. Source lineage overlap is explicitly accounted for in the local report. Both share public ancestors. Local match results cannot predict Kaggle ratings.
