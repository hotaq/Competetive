```python
@dataclass
class Settings:
    target: str
    identifier: str | None
    shuffle: bool
    splits: int
    seed:int
    path:str
    

    def __repr__(self) -> str:
        return f"""
        {"="*12} Settings {"="*12}
        ==> Target: {self.target}
        ==> Identifier: {self.identifier}
        ==> Shuffle: {self.shuffle}
        ==> Number of splits: {self.splits}
        ==> Seed: {self.seed}
        {"="*34}
        """

CFG = Settings(
    target="Will_Buy_EV",
    identifier="id",
    shuffle=True,
    splits=3,
    seed=152,
    path="/kaggle/input/competitions/playground-series-s6e9"
)

np.random.seed(CFG.seed)

CFG
```

```python
def read_data(directory:str) -> tuple:
    """
        Read CSV files from a directory and return train, test, and submission DataFrames.
    """
    data = {}
    for filename in os.listdir(directory):
        key = re.search("(.*).csv", filename).group(1)
        data[key] = pd.read_csv(
            os.path.join(
                directory,
                filename
            )
        )

    train = data["train"]
    test  = data["test"]
    sub   = data["sample_submission"]

    return train, test, sub
```

```python
def plot_kde(df1: pd.DataFrame,
             df2: pd.DataFrame,
             cols: int = 3,
             fill: bool = True,
             figsize: tuple = (15, 12),
             colors: list = ["#009E73", "#D55E00"],
             title: str = None) -> None:

    df1 = df1.copy()
    df2 = df2.copy()

    df1.drop([CFG.target], axis=1, inplace=True)

    numeric_columns = df1.select_dtypes(include='number').columns
    rows = np.ceil(len(numeric_columns) / cols).astype(int)

    fig, ax = plt.subplots(nrows=rows, ncols=cols, figsize=figsize)
    fig.patch.set_facecolor("#FFFCFC")

    for i, x in enumerate(numeric_columns):
        current_ax = ax[i // cols, i % cols] if rows > 1 else ax[i % cols]
        current_ax.set_facecolor("#FFFCFC")
        sns.kdeplot(x=x, color=colors[0], fill=fill, data=df1, edgecolor="black", label='Train', ax=current_ax, linewidth=2.5)
        sns.kdeplot(x=x, color=colors[1], fill=fill, data=df2, edgecolor="black", label='Test', ax=current_ax, linewidth=2.5)
        current_ax.set_xlabel(x, fontsize=14, weight="bold")
        current_ax.set_ylabel("Density", fontsize=12)
        current_ax.tick_params(labelsize=11)

        current_ax.spines["left"].set_visible(False)
        current_ax.spines["bottom"].set_visible(True)
        current_ax.spines["right"].set_visible(False)
        current_ax.spines["top"].set_visible(False)

        current_ax.spines["bottom"].set_color("#000")
        current_ax.spines["bottom"].set_linewidth(2)
        
        current_ax.xaxis.grid(False)
        current_ax.yaxis.grid(True, color="#000", linestyle="--", linewidth=1.0, alpha=0.2, zorder=0)
        current_ax.set_axisbelow(True)

    if len(numeric_columns) < (rows * cols):
        for i in range(len(numeric_columns), rows * cols):
            fig.delaxes(np.atleast_1d(ax).flatten()[i])

    legend_elements = [
        Patch(facecolor=colors[0], edgecolor='black', label='Train'),
        Patch(facecolor=colors[1], edgecolor='black', label='Test')
    ]

    fig.legend(handles=legend_elements, loc='upper center', ncol=2,
               bbox_to_anchor=(0.5, 0.97), frameon=False, fontsize=13)

    if title:
        fig.suptitle(title, fontsize=22, weight="bold", y=0.99)

    fig.subplots_adjust(left=0.07, right=0.97, top=0.90, bottom=0.07, hspace=0.5, wspace=0.3)
    plt.show()
```

```python
plot_kde(train, test, fill=True, title="Feature Distributions: Train vs Test")
```

<div style="background:#FEFCF3; border:1px solid #E8E0C8; border-left:5px solid #EAD98B; border-radius:12px; padding:22px 26px; margin:12px 0;">
<p style="margin:0 0 12px 0; font-family:'Courier New', Courier, monospace; font-size:12px; font-weight:bold; letter-spacing:3px; color:#8A7D4B;">WHY AM I DOING THIS?</p>
<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
Pearson correlation helps when we want to find linear relationships between variables. By "linear relationships" I mean that some random variable Y varies linearly with some variable Xi. If you remember the linear equation from school:
</p>
</div>

$$y = ax + b$$

<div style="background:#FEFCF3; border:1px solid #E8E0C8; border-left:5px solid #EAD98B; border-radius:12px; padding:22px 26px; margin:12px 0;">
<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
Where a is called the slope, or the inclination of the line, and b its y-intercept. If Y has a linear relationship with a set of variables X = {X1, ..., Xn}, then we can model Y as a linear combination of X:
</p>
</div>

$$Y = a_1 X_1 + \dots + a_n X_n + b$$

<div style="width:36px; height:3px; background:#EAD98B; margin:10px 0 0 0; border-radius:10px;"></div>

<div style="background:#FEFCF3; border:1px solid #E8E0C8; border-left:5px solid #EAD98B; border-radius:12px; padding:22px 26px; margin:12px 0;">
<p style="margin:0 0 12px 0; font-family:'Courier New', Courier, monospace; font-size:12px; font-weight:bold; letter-spacing:3px; color:#8A7D4B;">WHY AM I USING THIS?</p>
<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
I want to check whether each feature Xi has marginal dependence on the target Y. If Y depends on Xi, then there exists at least one value x such that:
</p>
</div>

$$P(Y|X_i = x) \neq P(Y)$$

<div style="background:#FEFCF3; border:1px solid #E8E0C8; border-left:5px solid #EAD98B; border-radius:12px; padding:22px 26px; margin:12px 0;">
<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
In other words, knowing x gives me information about the probabilities of Y. So, for example, if I have a random variable <em>Age</em> that has marginal dependence with <em>buying an EV car</em>, then it's possible that:
</p>
</div>

$$ P(\text{Buying an EV car} | \text{Age} = 21) \neq P(\text{Buying an EV car} | \text{Age} = 60) $$

<div style="background:#FEFCF3; border:1px solid #E8E0C8; border-left:5px solid #EAD98B; border-radius:12px; padding:22px 26px; margin:12px 0;">
<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
Note that this is an example. They can actually be the same. <strong>Dependence only requires that some value xi have the property mentioned above.</strong> For example, if P(Buying an EV car|Age=x) is equal for all possible age values from 21 to 60 and different for a single value over 60, then <em>Buying an EV car</em> is dependent on <em>Age</em>. On the other hand, if there is no dependence, then for every possible age it must hold that:
</p>
</div>

$$ P(\text{Buying an EV car} | \text{Age} = x) = P(\text{Buying an EV car})$$

<div style="background:#FEFCF3; border:1px solid #E8E0C8; border-left:5px solid #EAD98B; border-radius:12px; padding:22px 26px; margin:12px 0;">
<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
The probability of buying an EV doesn't change when I take age into account, so knowing a person's age doesn't help me determine whether they are more likely to buy an EV.
</p>
</div>

<div style="background:#FEFCF3; border:1px solid #E8E0C8; border-left:5px solid #EAD98B; border-radius:12px; padding:22px 26px; margin:12px 0;">

<p style="margin:0 0 12px 0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
Before, correlation allowed us to check for linear relationships between variables, while mutual information allowed us to assess the dependence between the target and the features. Now, we are asking a different question: <strong>If I add noise to a feature, how much does the model's performance suffer?</strong>
</p>
<p style="margin:0 0 12px 0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
I will use <code style="background:#F1EAD0; padding:2px 6px; border-radius:6px; font-size:14px;">permutation_importance</code> from <code style="background:#F1EAD0; padding:2px 6px; border-radius:6px; font-size:14px;">sklearn</code> to do this. <strong>If you want to implement it yourself, the idea is roughly:</strong>
</p>
<div style="background:#FFFFFF; border:1px solid #E8E0C8; border-radius:10px; padding:16px 20px; margin:0 0 12px 0;">
<p style="margin:0; font-family:Arial, sans-serif; font-size:15px; line-height:1.8; color:#2B2B2B;">
1. Split the data into train and validation sets.<br>
2. Train your model on the training set.<br>
3. Evaluate the model on the validation set and record the baseline score.<br>
4. For each feature in the validation set, randomly shuffle the values of that column.<br>
5. Evaluate the model again using the shuffled validation data.<br>
6. The permutation importance is the difference between the baseline score and the score after shuffling the feature.<br>
7. The closer the importance is to zero, the less evidence we have that the model relies on that feature for predictive performance.
</p>
</div>
<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
<strong>I will also add a random column as a negative control.</strong> I mean, if everything is working fine, then we expect that column to be really close to zero
</p>
</div>

```python
X["noise"] = np.random.randn(len(X))

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=CFG.seed)

clf = Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(random_state=CFG.seed))
]).fit(X_train, y_train)

result = permutation_importance(clf, X_test, y_test, scoring="roc_auc", n_repeats=30, random_state=CFG.seed)

imp = pd.DataFrame(result.importances.T, columns=X_test.columns)
order = imp.mean().sort_values(ascending=True).index.tolist()
imp = imp[order]
```

<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
<strong>Note:</strong> Using three models is totally arbitrary — or, as we say in Brazil, <em>tirado da bunda</em> — but in most Kaggle Playgrounds, the algorithms used tend to come from these families. Also, notice that this approach is tied to the order in which we remove columns.
```python
def validate_hypothesis(models, X, y, noise_cols, splits=5):
    X = X.copy()
    noise_cols = noise_cols.copy()
    
    results = defaultdict(list)

    console.print(f"[bold {ORANGE}]Baseline [/bold {ORANGE}]")
    for name, model in models.items():
        result = cross_val_score(model, X, y, cv=splits, scoring="roc_auc")
        results[name].append(result)
        print(f"{name}: {result.mean():.4f} ± {result.std():.4f}")

    while len(noise_cols) > 0:
        col = noise_cols.pop(0)
        X.drop(col, axis=1, inplace=True)

        console.print(f"\n[bold {STRONG_ORANGE}]Removed:[/bold {STRONG_ORANGE}] {col}")
        for name, model in models.items():
            result = cross_val_score(model, X, y, cv=splits, scoring="roc_auc")
            results[name].append(result)
            print(f"{name}: {result.mean():.4f} ± {result.std():.4f}")

    return results
```

Averaging predictions across seeds helps minimize seed variation and can lead to better predictions. Each additional seed adds running time, but more seeds give more stable estimates.
</p>

```python
def cross_validation(model, X, y, test, cv, n_seeds=3, to_encode=None, eval_set=False, bin_income=False):
    oofs = {}
    all_scores = []

    if to_encode is not None:
        encoded = [f"{c}_te" for c in to_encode]

    for seed in range(n_seeds):
        current_seed = CFG.seed + seed
        oof_train_seed = np.zeros(X.shape[0])
        oof_test_seed = np.zeros(test.shape[0])
        seed_scores = []

        cv.random_state = current_seed

        for i, (train_idx, val_idx) in enumerate(cv.split(X, y)):
            X_train, y_train = X.iloc[train_idx].copy(), y.iloc[train_idx]
            X_val, y_val = X.iloc[val_idx].copy(), y.iloc[val_idx]
            test_fold = test.copy()

            if bin_income:
                kb = KBinsDiscretizer(n_bins=600, encode="ordinal", strategy="quantile")
                X_train["inc_bin_600"] = kb.fit_transform(X_train[["Annual_Income_USD"]]).ravel().astype("int32")
                X_val["inc_bin_600"] = kb.transform(X_val[["Annual_Income_USD"]]).ravel().astype("int32")
                test_fold["inc_bin_600"] = kb.transform(test_fold[["Annual_Income_USD"]]).ravel().astype("int32")

            if to_encode is not None:
                te = TargetEncoder(target_type="binary", smooth="auto")
                X_train[encoded] = te.fit_transform(X_train[to_encode].astype(str), y_train)
                X_val[encoded] = te.transform(X_val[to_encode].astype(str))
                test_fold[encoded] = te.transform(test_fold[to_encode].astype(str))

            clf = clone(model)
            clf.random_state = current_seed

            if eval_set:
                clf.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)
            else:
                clf.fit(X_train, y_train)

            p_val = clf.predict_proba(X_val)[:, 1]
            p_test = clf.predict_proba(test_fold)[:, 1]

            oof_train_seed[val_idx] = p_val
            oof_test_seed += p_test / CFG.splits

            score = roc_auc_score(y_val, p_val)
            seed_scores.append(score)
            all_scores.append(score)

            console.print(f"[bold] Seed {current_seed} | Fold {i + 1}: {score:.4f} [/bold]")

        console.print(f"[bold {ORANGE}]Seed {current_seed}: {np.mean(seed_scores):.4f} ± {np.std(seed_scores):.4f}[/]")
        print()

        oofs[f"seed_{current_seed}"] = {
            "train": oof_train_seed,
            "test": oof_test_seed,
            "scores": seed_scores,
            "mean_score": np.mean(seed_scores)
        }

    console.print(f"[bold {STRONG_ORANGE}]Overall: {np.mean(all_scores):.4f} ± {np.std(all_scores):.4f}[/]")

    return oofs
```

<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;">
Notice that in the cell below I use <code style="background:#F1EAD0; padding:2px 6px; border-radius:6px; font-size:14px;">IsotonicRegression</code> from <code style="background:#F1EAD0; padding:2px 6px; border-radius:6px; font-size:14px;">sklearn</code>. It fits a non-decreasing function to the data. Why does it matter? ROC AUC only cares about the ranking of predictions, which means that if you apply any strictly increasing function to them, the score will not change. For example, suppose you have 0.90 and 0.89 in first and second place, respectively, ranked from most probable to least probable. Then, if you add 0.01 to everyone, they are still in first and second place, the third is still third, the fourth is still fourth, and so on... In that case, ROC AUC doesn't change.
    </p>

<p style="margin:0; font-family:Arial, sans-serif; font-size:15.5px; line-height:1.75; color:#2B2B2B;"><code style="background:#F1EAD0; padding:2px 6px; border-radius:6px; font-size:14px;">IsotonicRegression</code>, though, is only non-decreasing, not strictly increasing, so it's allowed to map two different predictions to the exact same value. Say, for example, that two predictions are slightly out of place due to some random noise. In that case, Isotonic Regression can smooth out the difference caused by the noise by mapping both predictions to the same value, creating a tie. Because ROC AUC treats ties as half-correct rather than fully incorrect, this can sometimes improve the metric.
</p>
</div>

```python
oofs_train = pd.DataFrame({seed: data["train"] for seed, data in oofs.items()})
oofs_test = pd.DataFrame({seed: data["test"] for seed, data in oofs.items()})

calibrators = {}

for col in oofs_train.columns:
    raw = oofs_train[col].to_numpy()
    calibrated = np.zeros_like(raw)

    for train_idx, val_idx in skf.split(raw, y):
        iso = IsotonicRegression(out_of_bounds="clip")
        iso.fit(raw[train_idx], y.iloc[train_idx])
        calibrated[val_idx] = iso.transform(raw[val_idx])

    oofs_train[col] = calibrated

    final_iso = IsotonicRegression(out_of_bounds="clip")
    final_iso.fit(raw, y)
    oofs_test[col] = final_iso.transform(oofs_test[col])
    calibrators[col] = final_iso
```
